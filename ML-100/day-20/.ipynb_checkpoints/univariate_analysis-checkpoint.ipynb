```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("train.csv")
```

```python
df.head()
```

# 1. Categorical Data

## a. Countplot

```python
sns.countplot(data=df, x="Survived")
df["Survived"].value_counts()
```

## b. Piechart

```python
df["Pclass"].value_counts().plot(kind="pie",autopct= "%.2f")
```

# 2. Numerical Data

## a. histogram

```python
import matplotlib.pyplot as plt
plt.hist(df["Age"])
```

## b. Distplot

```python
sns.distplot(df["Age"])
```

## c. BoxPLot

```python
sns.boxplot(df["Fare"])
```
