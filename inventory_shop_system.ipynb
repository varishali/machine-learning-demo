```python
from abc import ABC, abstractmethod
from datetime import datetime
```

```python
# ---------Abstraction---------
```

```python
class Product(ABC):
    
    """ abstraction base class for all product """
    
    def __init__(self, product_id, name, price, quantity):
        self._product_id = product_id  #encspsulation 
        self._name = name
        self._price = price
        self._quantity = quantity
        
    @abstractmethod  
    def category(self):
        pass
        
    # getters
    def get_id(self):
        return self._product_id
    def get_name(self):
        return self._name
    def get_price(self):
        return self._price
    def get_quantity(self):
        return self._quantity
        
    # setters    
    def set_price(self, price):
        if price < 0:
            print('price cannot be negative')
            return
        self._price = price
        
    def add_stock(self, qty):
        self._quantity += qty
        
    def reduce_stock(self, qty):
        if qty > self._quantity:
            print(f"Not enough stock for {self.name}!")
            return False
        self._quantity -= qty
        return True
        
    def total_value(self):
        return self._price * self._quantity

    def __str__(self):
        return (f"[{self._product_id}] {self._name} | Category : {self.category()}"
                f" | Price : {self._price} | Qty : {self._quantity} | Total  : {self.total_value()}")
```

```python
# inheritance + polymorphism

class Electronics(Product):
    def category(self):
        return "Electronics"

class Grocery(Product):
    def category(self):
        return "Grocery"

class Clothing(Product):
    def category(self):
        return "Clothing"
```

```python
# ---------------- SHOP / INVENTORY CLASS ----------------
class Shop:
    def __init__(self, shop_name):
        self.shop_name = shop_name
        self._products = {}     # product_id -> Product object
        self._sales_log = []

    def add_product(self, product):
        if product.get_id() in self._products:
            print("Product ID already exists! Adding stock instead.")
            self._products[product.get_id()].add_stock(product.get_quantity())
        else:
            self._products[product.get_id()] = product
            print(f"Added: {product.get_name()}")

    def remove_product(self, product_id):
        if product_id in self._products:
            removed = self._products.pop(product_id)
            print(f"Removed: {removed.get_name()}")
        else:
            print("Product not found!")

    def sell_product(self, product_id, qty):
        product = self._products.get(product_id)
        if not product:
            print("Product not found!")
            return
        if product.reduce_stock(qty):
            amount = product.get_price() * qty
            self._sales_log.append({
                "product": product.get_name(),
                "qty": qty,
                "amount": amount,
                "time": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
            })
            print(f"Sold {qty} x {product.get_name()} for {amount}")

    def restock_product(self, product_id, qty):
        product = self._products.get(product_id)
        if product:
            product.add_stock(qty)
            print(f"Restocked {qty} units of {product.get_name()}")
        else:
            print("Product not found!")

    def show_inventory(self):
        print(f"\n--- {self.shop_name} Inventory ---")
        if not self._products:
            print("No products in inventory.")
        for product in self._products.values():
            print(product)

    def total_inventory_value(self):
        return sum(p.total_value() for p in self._products.values())

    def show_sales_log(self):
        print("\n--- Sales Log ---")
        if not self._sales_log:
            print("No sales yet.")
        for sale in self._sales_log:
            print(f"{sale['time']} | {sale['product']} x{sale['qty']} = {sale['amount']}")

    def total_revenue(self):
        return sum(s["amount"] for s in self._sales_log)


# ---------------- DEMO / MAIN PROGRAM ----------------
def main():
    my_shop = Shop("City Mart")

    # Adding products (Polymorphism: same method, different object types)
    my_shop.add_product(Electronics("E01", "Laptop", 55000, 5))
    my_shop.add_product(Grocery("G01", "Rice (1kg)", 60, 100))
    my_shop.add_product(Clothing("C01", "T-Shirt", 350, 30))

    my_shop.show_inventory()

    print("\n--- Selling Products ---")
    my_shop.sell_product("G01", 10)
    my_shop.sell_product("E01", 2)
    my_shop.sell_product("C01", 5)

    print("\n--- Restocking ---")
    my_shop.restock_product("G01", 50)

    my_shop.show_inventory()
    my_shop.show_sales_log()

    print(f"\nTotal Inventory Value: {my_shop.total_inventory_value()}")
    print(f"Total Revenue: {my_shop.total_revenue()}")


if __name__ == "__main__":
    main()
```
