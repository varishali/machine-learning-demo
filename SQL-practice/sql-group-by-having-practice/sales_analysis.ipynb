```python
import pandas as pd
import sqlite3
```

```python
# read csv
df = pd.read_csv('sales_data_sample.csv',encoding='latin1')
```

```python
conn = sqlite3.connect(':memory:')
df.to_sql('sales', conn, index=False, if_exists='replace')

print('Data Load Ho Gaya')
print(df.columns.tolist())  #column ka name dekhne ke liye
```

```python
query = "SELECT * FROM sales LIMIT 5;"
pd.read_sql(query, conn)
```

```python
query = "SELECT DISTINCT PRODUCTLINE FROM sales;"
pd.read_sql(query, conn)
```

## == har product ka total sales ==

```python
query = "SELECT PRODUCTLINE, SUM(SALES) AS total_sales FROM sales GROUP BY PRODUCTLINE ORDER BY total_sales DESC;"
pd.read_sql(query, conn)
```

## == country bise order ==

```python
query = "SELECT COUNTRY, COUNT(*) AS total_orders FROM sales GROUP BY COUNTRY ORDER BY total_orders DESC;"
pd.read_sql(query, conn)
```

## == top 5 highest sales bale order ==

```python
query = "SELECT ORDERNUMBER, PRODUCTLINE, SALES FROM sales ORDER BY SALES DESC LIMIT 5;"
pd.read_sql(query, conn)
```

## == year bise total_sales ==

```python
query = "SELECT YEAR_ID, SUM(SALES) AS total_sales FROM sales GROUP BY YEAR_ID ORDER BY YEAR_ID;"
pd.read_sql(query, conn)
```

## == product jinka total 500000 se jyada ==

```python
query = "SELECT PRODUCTLINE, SUM(SALES) AS total_sales FROM sales GROUP BY PRODUCTLINE HAVING SUM(SALES) > 500000;"
pd.read_sql(query, conn)
```

## == cancelled order ==

```python
query = "SELECT COUNT(*) AS cancelled_orders FROM sales WHERE STATUS = 'Cancelled';"
pd.read_sql(query, conn)
```

## == har country ka total sale ==

```python
query  = "SELECT COUNTRY, SUM(SALES) AS total_sales FROM sales GROUP BY COUNTRY ORDER BY total_sales DESC;"
pd.read_sql(query, conn)
```

## == shiped order count ==

```python
query = "SELECT COUNT(*) AS shipped_order FROM sales WHERE STATUS = 'Shipped';"
pd.read_sql(query, conn)
```

## == har product ka avg quantity ==

```python
query = "SELECT PRODUCTLINE, AVG(QUANTITYORDERED) AS avg_qty FROM sales GROUP BY PRODUCTLINE"
pd.read_sql(query, conn)
```

## == AND condition - USA + large deals ==

```python
query = "SELECT * FROM sales WHERE COUNTRY ='USA' AND DEALSIZE ='Large';"
pd.read_sql(query, conn)
```

## == sales > 5000 , quantityorders > 40 ==

```python
query = "SELECT * FROM sales WHERE SALES > 5000 OR QUANTITYORDERED > 40;"
pd.read_sql(query, conn)
```

## == sari unique country ==

```python
query = "SELECT DISTINCT COUNTRY FROM sales"
pd.read_sql(query, conn)
```

## == car world name bala customer ==

```python
query = "SELECT DISTINCT CUSTOMERNAME FROM sales WHERE CUSTOMERNAME LIKE '%Car%';"
pd.read_sql(query, conn)
```
