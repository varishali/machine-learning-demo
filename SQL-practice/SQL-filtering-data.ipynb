```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect(':memory:')
```

```python
# create table
conn.execute("""
CREATE TABLE sales (
    id INTEGER,
    product TEXT,
    category TEXT,
    amount INTEGER,
    city TEXT,
    sale_date TEXT
);
""")
```

```python
conn.executemany("INSERT INTO sales VALUES (?,?,?,?,?,?)", [
    (1, 'Laptop', 'Electronics', 55000, 'Mumbai', '2024-01-05'),
    (2, 'Mouse', 'Electronics', 500, 'Mumbai', '2024-01-06'),
    (3, 'Shirt', 'Clothing', 1200, 'Delhi', '2024-01-07'),
    (4, 'Laptop', 'Electronics', 52000, 'Delhi', '2024-01-10'),
    (5, 'Jeans', 'Clothing', 2000, 'Mumbai', '2024-01-12'),
    (6, 'Mouse', 'Electronics', 450, 'Pune', '2024-01-15'),
    (7, 'Shirt', 'Clothing', 1100, 'Pune', '2024-01-18'),
])
```

```python
conn.commit()
print('Table Ready')
```

query = "SELECT * FROM sales;"

# filtering

```python
query = "SELECT * FROM sales WHERE city = 'Mumbai';"
pd.read_sql(query, conn)
```

```python
query = "SELECT * FROM sales WHERE amount > 50000;"
pd.read_sql(query, conn)
```

```python
query = "SELECT * FROM sales WHERE product = 'Mouse';"
pd.read_sql(query, conn)
```

## 3. text columns se sort

## 4. do cloumns se sorting

```python
query = "SELECT * FROM sales ORDER BY category, amount;"
pd.read_sql(query, conn)
```

## 5. group by ke result ko sort karna

```python
query = "SELECT category, SUM(amount) AS total_amount FROM sales GROUP BY category ORDER BY total_amount DESC;"
pd.read_sql(query, conn)
```

# HAR CITY KA TOTAL SALES

```python
query = "SELECT city, SUM(amount) AS total_amount FROM sales GROUP BY city"
pd.read_sql(query, conn)
```

# HAR PRODUCT KA AVERAGE AMOUNT

```python
query = "SELECT product, AVG(amount) AS total_amount FROM sales GROUP BY product"
pd.read_sql(query, conn)
```

# HAR COTEGORY KE TOTAL ORDER, HIGHEST SE LOWEST

```python
query = "SELECT category, COUNT(*) AS total_orders FROM sales GROUP BY category ORDER BY total_orders DESC;"
pd.read_sql(query, conn)
```

# HAR CITY MAIN SABSE BADI SINGLE SALE

```python
query = "SELECT city, MAX(amount) AS max_sale FROM sales GROUP BY city"
pd.read_sql(query, conn)
```

# CATEGORY JINKA TOTAL 1500 SE JYADA HAI

```python
query = "SELECT category, SUM(amount) AS total_amount FROM sales GROUP BY category HAVING SUM(amount) > 1500 DESC;"
pd.read_sql(query, conn)
```
